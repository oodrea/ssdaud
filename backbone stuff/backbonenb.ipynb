### generate anchor config

```python
import pandas as pd
import math

# Function to compute strides for anchor config
def compute_strides(input_size, feature_map_sizes):
    strides = [math.ceil(input_size / size) for size in feature_map_sizes] # use ceiling
    return strides

# Define input size and feature map sizes
input_size = 300  # Change this if needed
feature_map_sizes = [38, 19, 19, 19, 10, 10]  # Example feature map sizes

# Compute strides
strides = compute_strides(input_size, feature_map_sizes)

# Create a DataFrame for visualization
df = pd.DataFrame({
    "Stride": strides,
    "Feature Map Size": feature_map_sizes

})

# Display the DataFrame

display(df)
```

## MobileNet

### prints all layers in mobilenet

```python
import torch
from torchvision.models import mobilenet_v2

# Load the MobileNetV2 model
model = mobilenet_v2(weights="IMAGENET1K_V1")
model.eval()  # Set to evaluation mode

# Print all layers with their names
print("\nAll Layers in MobileNetV2:\n")
for name, layer in model.named_modules():
    print(f"{name} -> {layer}")

# Generate feature map resolutions
dummy_input = torch.randn(1, 3, 300, 300)  # Assuming input size is 300x300 for SSD
output_maps = {}

# Register hooks to capture feature maps
def hook_fn(module, input, output):
    output_maps[module.name] = output.shape  # Store output shape

# Attach hooks to all layers
for name, layer in model.named_modules():
    layer.name = name  # Assign layer name for reference
    layer.register_forward_hook(hook_fn)

# Forward pass
with torch.no_grad():
    model(dummy_input)

# Print feature map resolutions
print("\nFeature Map Resolutions:")
for name, shape in output_maps.items():
    print(f"{name}: {shape}")  # (Batch, Channels, Height, Width)
```
